```python
import pandas as pd

df = pd.read_csv(os.path.expanduser("~/Downloads/tested.csv"))

df.head()
```

```python
print("Median Fare:", df['Fare'].median())

df['Fare'] = df['Fare'].fillna(df['Fare'].median())

print(df)
```

```python
print(df['Fare'].isnull().sum())
```

```python
data = {
    'Student': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena'],
    'Department': ['CSE', 'ECE', None, 'ISE', None]
}

df2 = pd.DataFrame(data)

print("Before Forward Fill:")
print(df2)

df2['Department'] = df2['Department'].ffill()

print("\nAfter Forward Fill:")
print(df2)
```
